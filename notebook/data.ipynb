# Cleaning Data

```python
import numpy as np

# Buang kolom homepage
movies = movies.drop(columns=['homepage'])

# Isi tagline kosong
movies['tagline'] = movies['tagline'].fillna('No Tagline')

# Isi overview kosong
movies['overview'] = movies['overview'].fillna('No Overview')

# ubah release_date menjadi datetime dan buang baris dengan release_date kosong
movies['release_date'] = pd.to_datetime(movies['release_date'], errors='coerce')
movies = movies.dropna(subset=['release_date'])

# Isi runtime kosong dengan median
movies['runtime'] = movies['runtime'].replace(0, np.nan)
movies['runtime'] = movies['runtime'].fillna(movies['runtime'].median())

# Ubah budget dan revenue bernilai 0 jadi dianggap kosong
movies['budget'] = movies['budget'].replace(0, np.nan)
movies['revenue'] = movies['revenue'].replace(0, np.nan)

# Ubah vote_average menjadi NaN jika vote_count = 0
movies.loc[
    movies['vote_count'] == 0,
    'vote_average'
] = np.nan
```

# Verifikasi Hasil Cleaning

```python
print("Missing values setelah cleaning:")
print(movies.isnull().sum())

print("Jumlah baris setelah cleaning:", len(movies))
print("Jumlah kolom setelah cleaning:", len(movies.columns))

print("Duplikat movies setelah cleaning:",
      movies.duplicated().sum())

print("Film dengan runtime 0 menit:", (movies['runtime'] == 0).sum())
print("Film dengan vote_count = 0 tapi vote_average > 0:", ((movies['vote_count'] == 0) & (movies['vote_average'] > 0)).sum())
print("Film dengan vote_average=0 & vote_count=0 (belum terperbaiki):", ((movies['vote_average'] == 0) & (movies['vote_count'] == 0)).sum())
```

# Deteksi Outlier

```python
for col in numeric_cols:
    Q1 = movies[col].quantile(0.25)
    Q3 = movies[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((movies[col] < lower) | (movies[col] > upper)).sum()
    
    print(f"{col}: {outliers} outliers")
```

# Data Integration

```python
df = movies.merge(credits, 
                  left_on='id', 
                  right_on='movie_id', 
                  how='inner')

print("Shape sebelum integrasi:", movies.shape)
print("Shape setelah integrasi:", df.shape)

# Cek redundansi kolom title
print("Redundansi kolom title:", (df['title_x'] == df['title_y']).all())
```

```python
# Rapikan
df = df.drop(columns=['title_y', 'movie_id'])
df = df.rename(columns={'title_x': 'title'})
```

```python
import ast


# Fungsi universal yang aman untuk mengekstrak nama dari string JSON
def ekstrak_nama(teks, limit=None):
    try:
        if pd.isna(teks) or teks == "":
            return []
        daftar = ast.literal_eval(teks)
        hasil = [item['name'] for item in daftar if 'name' in item]
        return hasil[:limit] if limit is not None else hasil
    except (ValueError, SyntaxError, TypeError):
        return []

# Fungsi khusus untuk mencari Sutradara di kolom crew
def ambil_sutradara(teks):
    try:
        if pd.isna(teks) or teks == "":
            return None
        daftar = ast.literal_eval(teks)
        for item in daftar:
            if item.get('job') == 'Director':
                return item.get('name')
        return None
    except (ValueError, SyntaxError, TypeError):
        return None

```

```python

# Ekstraksi Fitur & Pembuatan Kolom Angka Baru (Sesuai Rencana)

# 1. Bongkar data teks / list
df['genres_list'] = df['genres'].apply(ekstrak_nama)
df['cast_list'] = df['cast'].apply(lambda x: ekstrak_nama(x, limit=3)) # 3 Pemain utama
df['director'] = df['crew'].apply(ambil_sutradara)

# 2. Tambahkan kolom angka baru untuk PCA (menambah 2 fitur numerik baru)
df['genre_count'] = df['genres_list'].apply(len)
df['cast_count'] = df['cast'].apply(lambda x: len(ekstrak_nama(x))) # Total seluruh cast asli

# 3. Ekstraksi opsional (Hanya disimpan untuk insight tambahan / EDA terpisah, tidak masuk PCA)
df['keywords_list'] = df['keywords'].apply(ekstrak_nama)
df['production_companies_list'] = df['production_companies'].apply(ekstrak_nama)

```

```python
# Menyiapkan data numerik khusus untuk PCA
df_pca_input = df[numeric_cols].copy()

# Isi missing value dengan median
df_pca_input = df_pca_input.fillna(
    df_pca_input.median()
)

print("Jumlah NaN untuk input PCA:")
print(df_pca_input.isnull().sum())
```

# Standaridisasi (Z-score)

```python
from sklearn.preprocessing import StandardScaler

# Pastikan tidak ada NaN lagi sebelum standardisasi
print("Cek NaN sebelum scaling:")
print(df_pca_input.isnull().sum())

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_pca_input)
df_scaled = pd.DataFrame(df_scaled, columns=df_pca_input.columns, index=df_pca_input.index)

df_scaled.head()
```

```python
df_scaled.describe() 
# mean = 0, std = 1 berarti data sudah dinormalisasi dengan benar.
```

```python
from sklearn.decomposition import PCA

pca = PCA()
pca_result = pca.fit_transform(df_scaled)
```

```python
print("Explained variance ratio per komponen:")
print(pca.explained_variance_ratio_)
```

```python
# Hitung Kumulatif
print("Kumulatif variance:", pca.explained_variance_ratio_.cumsum())

for i, variance in enumerate(pca.explained_variance_ratio_.cumsum(), start=1):
    print(f"PC1-PC{i}: {variance:.2%}")
```

```python
pca_final = PCA(n_components=4)
pca_result_final = pca_final.fit_transform(df_scaled)

df_pca = pd.DataFrame(pca_result_final, columns=['PC1', 'PC2', 'PC3', 'PC4'], index=df.index)
df_pca['title'] = df['title'].values

print("Total variance terjaga dengan 4 komponen:", pca_final.explained_variance_ratio_.sum())
df_pca.head()
```

```python
cumulative_variance = pca.explained_variance_ratio_.cumsum()

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='o'
)
plt.axhline(y=0.80, color='r', linestyle='--', label='Batas 80%')
plt.axvline(x=4, color='g', linestyle='--', label='4 komponen dipilih')
plt.xlabel('Jumlah Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance PCA')
plt.xticks(range(1, len(cumulative_variance) + 1))
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# %%
plt.figure(figsize=(8, 6))
plt.scatter(df_pca['PC1'], df_pca['PC2'], alpha=0.3, s=10)
plt.xlabel(f'PC1 ({pca_final.explained_variance_ratio_[0]*100:.1f}% variance)')
plt.ylabel(f'PC2 ({pca_final.explained_variance_ratio_[1]*100:.1f}% variance)')
plt.title('Hasil PCA: Proyeksi Film ke 2 Komponen Utama')
plt.tight_layout()
plt.show()
```

```python
# PCA Loadings
loadings = pd.DataFrame(
    pca_final.components_.T,
    columns=['PC1', 'PC2', 'PC3', 'PC4'],
    index=numeric_cols
)

display(loadings)
```

```python
# Data bermasalah SEBELUM = NaN + nilai 0 yang tidak valid
sebelum = pd.Series({
    'homepage': movies_raw['homepage'].isnull().sum(),
    'tagline': movies_raw['tagline'].isnull().sum(),
    'overview': movies_raw['overview'].isnull().sum(),
    'release_date': movies_raw['release_date'].isnull().sum(),
    'runtime': movies_raw['runtime'].isnull().sum() + (movies_raw['runtime'] == 0).sum(),
    'budget': (movies_raw['budget'] == 0).sum(),
    'revenue': (movies_raw['revenue'] == 0).sum(),
    'vote_average': (movies_raw['vote_count'] == 0).sum(),
})

def hitung_nan(data, k):
    return data[k].isnull().sum() if k in data.columns else 0

def hitung_nan_pca(k):
    if k in df_pca_input.columns:
        return df_pca_input[k].isnull().sum()
    return hitung_nan(df, k)

tabel = pd.DataFrame({
    'Sebelum': sebelum,
    'Sesudah cleaning': pd.Series({k: hitung_nan(df, k) for k in sebelum.index}),
    'Sesudah imputasi (input PCA)': pd.Series({k: hitung_nan_pca(k) for k in sebelum.index}),
})
print(tabel)

besar = tabel[tabel['Sebelum'] >= 100]
kecil = tabel[tabel['Sebelum'] < 100]

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
besar.plot(kind='bar', ax=axes[0])
axes[0].set_title('Kolom dengan banyak data bermasalah')
axes[0].set_ylabel('Jumlah baris')
kecil.plot(kind='bar', ax=axes[1])
axes[1].set_title('Kolom dengan sedikit data bermasalah')
axes[1].set_ylabel('Jumlah baris')
fig.suptitle('Perbandingan Data Bermasalah Sebelum dan Sesudah Preprocessing',
             fontsize=13)
for ax in axes:
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()

```

```python

# Jumlah outlier sebelum (raw) vs sesudah cleaning
def hitung_outlier(data, cols):
    hasil = {}
    for col in cols:
        Q1, Q3 = data[col].quantile(0.25), data[col].quantile(0.75)
        IQR = Q3 - Q1
        hasil[col] = int(((data[col] < Q1 - 1.5*IQR) | (data[col] > Q3 + 1.5*IQR)).sum())
    return hasil

cols6 = ["budget", "revenue", "popularity", "runtime", "vote_average", "vote_count"]
outlier_table = pd.DataFrame({
    'Sebelum': hitung_outlier(movies_raw, cols6),
    'Sesudah': hitung_outlier(df, cols6),
})
print(outlier_table)
```
